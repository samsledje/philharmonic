```python
import glob
import pathlib
import re

import numpy as np
from scipy.stats import entropy

RESULTS_DIR = pathlib.Path(
    "/mnt/home/ssledzieski/Projects/philharmonic-experiments/results/p_damicornis/20240917-pdamAll-phil_v0.3.0/"
)
FILE_NAME_RE = re.compile(f"{RESULTS_DIR}/{RESULTS_DIR.name}_(.*)_clusters.txt")

clustering_files = glob.glob(str(RESULTS_DIR / "*clusters.txt"))

clusters = {}
for cf in clustering_files:
    cluster_type = re.match(FILE_NAME_RE, cf).groups()[0]
    with open(cf) as f:
        for line in f:
            cluster = line.strip().split()
            if cluster_type not in clusters:
                clusters[cluster_type] = []
            clusters[cluster_type].append(cluster)


def normalize(distribution: list):
    distribution = np.array(distribution)
    return distribution / distribution.sum()


cluster_sizes = {k: [len(c) for c in v] for k, v in clusters.items()}
cluster_sizes_as_distribution = {k: normalize(v) for k, v in cluster_sizes.items()}

CLUSTERS = [
    "eigenvector",
    "infomap",
    "ipca",
    "label_propagation",
    "leiden",
    "louvain",
    "manual_no_recipe",
    "manual",
    "multicom",
    "r_spectral_clustering",
]

# print sorted by maximum entropy
for k, v in sorted(
    [
        (k, entropy(v))
        for k, v in cluster_sizes_as_distribution.items()
        if k in CLUSTERS
    ],
    key=lambda x: x[1],
    reverse=True,
):
    print(f"{k}: {v}")
```
